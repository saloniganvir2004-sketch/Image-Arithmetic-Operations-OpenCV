```python
# ==========================================
# Image Addition using OpenCV (Google Colab)
# ==========================================

# Step 1: Import Libraries
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Step 2: Upload Two Images
print("Upload the Apple Image")
uploaded1 = files.upload()
img1_name = list(uploaded1.keys())[0]

print("\nUpload the Orange Image")
uploaded2 = files.upload()
img2_name = list(uploaded2.keys())[0]

# Step 3: Load the Images
img1 = cv2.imread(img1_name)
img2 = cv2.imread(img2_name)

# Step 4: Resize the Images
img1 = cv2.resize(img1, (400, 400))
img2 = cv2.resize(img2, (400, 400))

# Step 5: Convert Images to Grayscale
gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

# Step 6: Add the Two Images
result = cv2.add(gray1, gray2)

# Step 7: Display the Images
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(gray1, cmap='gray')
plt.title("Apple (Grayscale)")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(gray2, cmap='gray')
plt.title("Orange (Grayscale)")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(result, cmap='gray')
plt.title("Addition of Apple + Orange")
plt.axis("off")

plt.show()
```

```python
# ==========================================
# Image Multiplication using OpenCV (Google Colab)
# ==========================================

# Step 1: Import Libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Step 2: Upload Two Images
print("Upload the Apple Image")
uploaded1 = files.upload()
img1_name = list(uploaded1.keys())[0]

print("\nUpload the Vegetable Image")
uploaded2 = files.upload()
img2_name = list(uploaded2.keys())[0]

# Step 3: Load the Images
img1 = cv2.imread(img1_name)
img2 = cv2.imread(img2_name)

# Step 4: Resize the Images
img1 = cv2.resize(img1, (400, 400))
img2 = cv2.resize(img2, (400, 400))

# Step 5: Convert Images to Grayscale
gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

# Step 6: Multiply the Two Images
# Convert to float to avoid overflow
gray1 = gray1.astype(np.float32) / 255.0
gray2 = gray2.astype(np.float32) / 255.0

result = gray1 * gray2

# Convert back to uint8
result = (result * 255).astype(np.uint8)

# Step 7: Display the Images
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(gray1, cmap='gray', vmin=0, vmax=1)
plt.title("Apple (Grayscale)")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(gray2, cmap='gray', vmin=0, vmax=1)
plt.title("Vegetable (Grayscale)")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(result, cmap='gray')
plt.title("Multiplication of Apple × Vegetable")
plt.axis("off")

plt.show()
```

```python
# ==========================================
# Image Multiplication using OpenCV (Google Colab)
# ==========================================

# Step 1: Import Libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Step 2: Upload Two Images
print("Upload the Apple Image")
uploaded1 = files.upload()
img1_name = list(uploaded1.keys())[0]

print("\nUpload the Vegetable Image")
uploaded2 = files.upload()
img2_name = list(uploaded2.keys())[0]

# Step 3: Load the Images
img1 = cv2.imread(img1_name)
img2 = cv2.imread(img2_name)

# Step 4: Resize the Images
img1 = cv2.resize(img1, (400, 400))
img2 = cv2.resize(img2, (400, 400))

# Step 5: Multiply the Two Color Images
# Convert to float to avoid overflow
img1 = img1.astype(np.float32) / 255.0
img2 = img2.astype(np.float32) / 255.0

result = img1 * img2

# Convert back to uint8
result = (result * 255).astype(np.uint8)

# Convert BGR to RGB for displaying with Matplotlib
img1_rgb = cv2.cvtColor((img1 * 255).astype(np.uint8), cv2.COLOR_BGR2RGB)
img2_rgb = cv2.cvtColor((img2 * 255).astype(np.uint8), cv2.COLOR_BGR2RGB)
result_rgb = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)

# Step 6: Display the Images
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(img1_rgb)
plt.title("Apple")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(img2_rgb)
plt.title("Vegetable")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(result_rgb)
plt.title("Multiplication of Apple × Vegetable")
plt.axis("off")

plt.show()
```

```python
# ==========================================
# Merge Two Images (Color) using OpenCV
# Google Colab
# ==========================================

# Step 1: Import Libraries
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Step 2: Upload Two Images
print("Upload the Apple Image")
uploaded1 = files.upload()
img1_name = list(uploaded1.keys())[0]

print("\nUpload the Vegetable Image")
uploaded2 = files.upload()
img2_name = list(uploaded2.keys())[0]

# Step 3: Load the Images
img1 = cv2.imread(img1_name)
img2 = cv2.imread(img2_name)

# Step 4: Resize the Images
img1 = cv2.resize(img1, (400, 400))
img2 = cv2.resize(img2, (400, 400))

# Step 5: Merge (Blend) the Two Images
# 50% Apple + 50% Vegetable
merged = cv2.addWeighted(img1, 0.5, img2, 0.5, 0)

# Step 6: Convert BGR to RGB for Display
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2_rgb = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
merged_rgb = cv2.cvtColor(merged, cv2.COLOR_BGR2RGB)

# Step 7: Display the Images
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(img1_rgb)
plt.title("Apple")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(img2_rgb)
plt.title("Vegetable")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(merged_rgb)
plt.title("Merged Image")
plt.axis("off")

plt.show()
```
